```python
import pandas as pd
```

```python
X = pd.read_csv('/content/train.csv')
y = pd.read_csv('/content/test.csv')

X.head()
```

```python
pixel_cols = [col for col in X.columns if 'pixel' in col]
cnt = 0
for col in pixel_cols:
    unique_vals = X[col].unique()
    print(f"Column: {col}")
    print(f"Unique values: {unique_vals}")

    # Check if all values are between 0 and 255
    is_within_range = (X[col] >= 0).all() and (X[col] <= 255).all()
    if(is_within_range == True): cnt += 1
    print(f"Are all values between 0 and 255?: {is_within_range}")
    print("\n")

print(cnt)
```

```python
# Normalize pixel values from 0-255 to 0-1

# Identify pixel columns (excluding 'label' if it exists in X)
pixel_cols = [col for col in X.columns if 'pixel' in col]

# Normalize training data (X)
X_normalized = X.copy()
for col in pixel_cols:
    X_normalized[col] = X_normalized[col].astype('float32') / 255.0

# Normalize test data (y)
y_normalized = y.copy()
for col in pixel_cols:
    y_normalized[col] = y_normalized[col].astype('float32') / 255.0

# Display the head of the normalized DataFrames
print("Normalized X (Training Data) Head:")
display(X_normalized.head())

print("\nNormalized y (Test Data) Head:")
display(y_normalized.head())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
from sklearn.tree import DecisionTreeClassifier

# Separate features (X_data) and labels (y_labels) from the normalized training data
X_data = X_normalized.drop('label', axis=1)
y_labels = X_normalized['label']

# Split the training data into training and validation sets
X_train, X_val, y_train, y_val = train_test_split(X_data, y_labels, test_size=0.2, random_state=42)

models = {
    "tree": DecisionTreeClassifier(),
    "svm_linear": SVC(kernel="linear"),
    "svm_rbf": SVC(kernel="rbf")
}

for name, model in models.items():
    # Fit models on the training set
    model.fit(X_train, y_train)
    # Evaluate models on the validation set (X_val, y_val)
    print(f"{name} validation accuracy: {model.score(X_val, y_val)}")
```

```python
from sklearn.metrics import confusion_matrix, classification_report

for name, model in models.items():
    print(f"\n--- Evaluation for {name.upper()} Model ---")

    # Make predictions on the validation set
    y_pred = model.predict(X_val)

    # Calculate and print Confusion Matrix
    conf_matrix = confusion_matrix(y_val, y_pred)
    print("Confusion Matrix:\n", conf_matrix)

    # Calculate and print Classification Report
    class_report = classification_report(y_val, y_pred)
    print("Classification Report:\n", class_report)
```

```python
print("\n--- Support Vector Analysis ---")

svm_linear_model = models["svm_linear"]
svm_rbf_model = models["svm_rbf"]

print("Linear SVM Support Vectors Shape:", svm_linear_model.support_vectors_.shape)
print("Number of Support Vectors per class (Linear):", svm_linear_model.n_support_)

print("\nRBF SVM Support Vectors Shape:", svm_rbf_model.support_vectors_.shape)
print("Number of Support Vectors per class (RBF):", svm_rbf_model.n_support_)
```

```python
print("\n--- Multiclass Strategy ---")
print("Linear SVM decision function shape:", svm_linear_model.decision_function(X_val[:5]).shape)
```

```python
from sklearn.decomposition import PCA

print("\n--- PCA Experiment ---")

# Reduce dimensionality
pca = PCA(n_components=50)

X_train_pca = pca.fit_transform(X_train)
X_val_pca = pca.transform(X_val)

# Train SVM again on reduced data
svm_pca = SVC(kernel='rbf')

svm_pca.fit(X_train_pca, y_train)

pca_accuracy = svm_pca.score(X_val_pca, y_val)

print("SVM Accuracy after PCA:", pca_accuracy)
```

PCA reduces dimensionality while preserving most important variance in the data. Even after reducing features from 784 to 50, the SVM model maintains high accuracy, demonstrating that many pixel features are redundant.

| Model         | Accuracy |
| ------------- | -------- |
| Decision Tree | 0.85     |
| Linear SVM    | 0.93     |
| RBF SVM       | 0.97     |
| RBF + PCA     | 0.978    |


The RBF kernel maps the input data into a higher dimensional feature space where nonlinear patterns become linearly separable. Instead of explicitly computing this transformation, the kernel trick allows SVM to compute similarities between data points efficiently.

The nonlinear RBF kernel requires more support vectors to define complex decision boundaries compared to linear SVM.
